## Frequency

Given a sample data set $X$, may it be numeric or categorical, the frequency
of a given value,

$$
freq(val)=\text{number of times a particular value appears in }X
$$


```python
def valores_frecuencias(x):
    """Finds unique values and their frequencies, returning sorted pairs."""
    # np.unique naturally sorts the 'values' array it returns
    values, counts = np.unique(x, return_counts=True)

    # zip them into pairs and convert to a list
    frequency_pairs = list(zip(values, counts))

    return frequency_pairs

# Example usage:
#   commute_A = [25, 10, 15, 20, 10, 15, 10]
#   result = valores_frecuencias(commute_A)
#   print(result)
# Output: [(10, 3), (15, 2), (20, 1), (25, 1)]
```

```python
#
# dotplots.        https://proclusacademy.com/blog/practical/dot-plot-python-matplotlib/#:~:text=And%20draw%20a%20dot%20plot,values%20on%20x%2Daxis.%20%23
#
def dotplot(input_x, **args):

    # Count how many times does each value occur
    unique_values, counts = np.unique(input_x, return_counts=True)

    # Convert 1D input into 2D array
    scatter_x = [] # x values
    scatter_y = [] # corresponding y values
    for idx, value in enumerate(unique_values):
        for counter in range(1, counts[idx]+1):
            scatter_x.append(value)
            scatter_y.append(counter)

    # draw dot plot using scatter()
    plt.scatter(scatter_x, scatter_y, **args)

    # Optional - show all unique values on x-axis.
    # Matplotlib might hide some of them
    plt.gca().set_xticks(unique_values)


    #EXAMPLE OF DOTPLOTS
hs_heights = np.array([
    71, 67, 64, 72, 65, 69, 66, 68, 69, 72,
    69, 73, 69, 72, 73, 74, 76, 68, 66, 63,
    67, 71, 72, 74, 68, 69, 75, 71, 72, 72,
    65, 66, 72, 74, 66, 62, 75, 75, 64, 63,
    64, 66, 74, 67, 72, 70, 71, 70, 74, 68
])

# Draw dot plot using our new function
dotplot(input_x=hs_heights)
```

# Centrality Estimators

Given a sample numeric data set $X=\{x_1,x_2,...,x_n\}$, assumed to be ordered (sorted) there a three centrality estimators,

## Mean

$$
\bar{x}=\frac{1}{n}\sum_{i=1}^n\,x_i
$$

## Mode

$$
mode(X)=\text{Element of }X\text{ having the highest frequency}$$

and,

## Median

the value that sits exaclty at the midle of $X$

$$
\tilde{x} =
\begin{cases}
x_{\left(\frac{n+1}{2}\right)} & \text{if } n \text{ is odd} \\[10pt]
\frac{1}{2} \left( x_{\left(\frac{n}{2}\right)} + x_{\left(\frac{n}{2} + 1\right)} \right) & \text{if } n \text{ is even}
\end{cases}
$$



```python
# Mean Value
def mean(x):
    """Calculates the average of a numerical list."""
    s = 0.0
    # Everything below is indented to stay inside the function
    for m in range(len(x)):
        s = s + x[m]

    s = s / len(x)
    return s
```

```python
def calculate_mode(x):
    # 1. Get unique values and their counts
    values, counts = np.unique(x, return_counts=True)

    # 2. Find the index of the largest count
    max_index = np.argmax(counts)

    # 3. Return the value at that index
    return values[max_index]

# Example:
commute_A = [15, 20, 15, 25, 20, 15]
print(calculate_mode(commute_A))  # Output: 15
```

```python
def manual_median(x):
    temp = sorted(x)
    n = len(temp)
    mid = n // 2

    if n % 2 == 0:  # Even number of items
        return (temp[mid - 1] + temp[mid]) / 2
    else:           # Odd number of items
        return temp[mid]
```

# Measures of spread

## Range

Given a numeric data set $X=\{x_1,x_2,...,x_m\}$

$$
Range=x_{Max}-x_{Min}
$$

```python
def s_range(x):
    """Calculates the average of a numerical list."""
    # x[-1] is the same as x[len(x)-1]
    R=x[-1]-x[0]
    return R
```

```python
# Sample variance
def sample_var(x):
    meanx=mean(x)
    v=0.0
    for m in range(len(x)):
        v=v+(x[m]-meanx)**2

    v=v/(len(x)-1)
    return v
```

```python
# Sample standard deviation
def sample_sdv(x):
SDV=np.sqrt(sample_var(x))
return SDV
```
